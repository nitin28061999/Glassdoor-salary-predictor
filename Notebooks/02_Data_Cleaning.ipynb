```python
df['Salary Estimate'] = df['Salary Estimate'].str.replace('(Glassdoor est.)','', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('$','', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('K','', regex=False)
```

```python
df['Salary Estimate'].unique()[:20]
```

```python
df['Salary Estimate'].sample(20)
```

```python
# Remove unwanted text
salary = df['Salary Estimate'].str.replace(r'\(Glassdoor est\.\)', '', regex=True)
salary = salary.str.replace(r'Employer Provided Salary:', '', regex=True)
salary = salary.str.replace('$', '', regex=False)
salary = salary.str.replace('K', '', regex=False)
salary = salary.str.replace(',', '', regex=False)

# Remove rows with -1
salary = salary.replace('-1', np.nan)

# Extract salary numbers
salary = salary.str.extract(r'(\d+)\s*-\s*(\d+)')

# Convert to numeric
df['Min Salary'] = pd.to_numeric(salary[0], errors='coerce')
df['Max Salary'] = pd.to_numeric(salary[1], errors='coerce')

# Average salary
df['Average Salary'] = (df['Min Salary'] + df['Max Salary']) / 2
```
